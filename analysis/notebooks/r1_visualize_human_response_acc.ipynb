```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
analysis_results = np.load("./output/r1_analysis_results.npy", allow_pickle=True).item()
```

```python
for acc, ttc, q_type in zip(analysis_results['acc'], analysis_results['ttc'], analysis_results['q_type']):
    plt.scatter(1-acc, ttc, color='blue')
    plt.text(1-acc, ttc, q_type)

plt.xlabel('Error Rate')
plt.grid(visible=True)
plt.ylabel('TTC in Seconds')
```

```python
for acc, tta, q_type in zip(analysis_results['acc'], analysis_results['tta'], analysis_results['q_type']):
    plt.scatter(1-acc, tta, color='blue')
    plt.text(1-acc, tta, q_type)

plt.xlabel('Error Rate')
plt.grid(visible=True)
plt.ylabel('TTA in Seconds')
```
